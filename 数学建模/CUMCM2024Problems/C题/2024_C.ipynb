```python
# 附件二数据预处理：计算平均销售单价，以及每种作物的预期销售量
df3 = pd.read_excel('附件2.xlsx',sheet_name = '2023年统计的相关数据')

# 计算销售单价的中点值
def calculate_midpoint_price(price_range):
    """计算价格区间的中点值"""
    if pd.isna(price_range) or price_range == '':
        return np.nan
    
    # 处理价格区间字符串，格式如 "2.50-4.00"
    try:
        # 分割价格区间
        parts = str(price_range).split('-')
        if len(parts) == 2:
            min_price = float(parts[0])
            max_price = float(parts[1])
            # 计算中点值
            midpoint = (min_price + max_price) / 2
            return midpoint
        else:
            return np.nan
    except:
        return np.nan
    
# 计算销售单价中点值
df3['销售单价中点值/(元/斤)'] = df3['销售单价/(元/斤)'].apply(calculate_midpoint_price)
df3
```
